## 벡터화 연산과 브로드캐스팅: 반복문 없는 고속 연산

```python
import numpy as np
# 벡터화 연산 예시
lst = [1, 2, 3]
lst_new = [x * 10 for x in lst] # ①
arr = np.array([1, 2, 3])
arr_new = arr * 10 # ②
print(arr_new)
# 브로드캐스팅 예시
a = np.array([1, 2, 3])
b = 5
result = a + b # ③
print(result)
matrix = np.array([[1, 2, 3], [4, 5, 6]]) # ④
vec = np.array([10, 20, 30])
result = matrix + vec # ⑤
print(result) 
```

```python
print("1. (5,3) + (3,) → 성공")
a1 = np.ones((5,3))
b1 = np.array([10, 20, 30])
print((a1 + b1).shape) # ①
print("2. (4,1,6) + (1,6) → 성공")
a2 = np.ones((4,1,6))
b2 = np.array([1,2,3,4,5,6])[np.newaxis, :] # ②
print((a2 + b2).shape) # ③
print("3. (7,1) + (7,) → 성공")
a3 = np.ones((7,1))
b3 = np.arange(7)
print((a3 + b3).shape) # ④
print("실패 사례: (2,3) + (2,)")
try:
    a4 = np.ones((2,3))
    b4 = np.array([10, 20])
    print(a4 + b4)
except ValueError as e:
    print("ValueError:", e) # ⑤
```

# 단계별 실습: ndarray 생성과 기본 연산

## 배열 생성: np.array(), np.arange(), np.linspace(), zeros, ones, full, eye

###단계 1: np.array()로 기본 배열 생성


```python
import numpy as np
lst_1d = [1, 2, 3, 4]
arr_1d = np.array(lst_1d) # ①
print(arr_1d)
print(arr_1d.shape)
lst_2d = [[1, 2, 3], [4, 5, 6]] # ②
arr_2d = np.array(lst_2d)
print(arr_2d)
print(arr_2d.shape) # ③ 
```

###단계 2: np.arange()와 np.linspace()로 수열 생성


```python
arr_range = np.arange(0, 10, 2) # ①
print(arr_range)
arr_lin = np.linspace(0, 1, 5)  # ②
print(arr_lin)
```

###단계 3: zeros, ones, full, eye로 특수 배열 생성


```python
import numpy as np
zeros_arr = np.zeros((2, 3))  # ①
print(zeros_arr)
ones_arr = np.ones((2, 3))    # ②
print(ones_arr)
full_arr = np.full((2, 2), 5) # ③
print(full_arr)
eye_arr = np.eye(3)           # ④
print(eye_arr)
```

## 데이터 타입 변환: astype()

```python
import numpy as np
arr = np.array([1.5, 2.5, 3.5]) # ①
print(arr.dtype)
arr_int = arr.astype(int) # ②
print(arr_int)
print(arr_int.dtype)
str_arr = np.array(['1', '2', '3']) # ③
int_str_arr = str_arr.astype(int)
print(int_str_arr) 
```

## 기본 연산: 사칙연산, 제곱, 제곱근, 비교 연산

###단계 1: 사칙연산 (+, -, , /), 제곱, 제곱근

```python
import numpy as np
a = np.array([1, 2, 3]) # ①
b = np.array([4, 5, 6])
# 덧셈, 뺄셈, 곱셈, 나눗셈
sum_ab = a + b
sub_ab = a - b
mul_ab = a * b
div_ab = a / b # ②
print(sum_ab)
print(sub_ab)
print(mul_ab)
print(div_ab)
# 제곱, 제곱근
sq_arr = np.square(a) # ③
sqrt_arr = np.sqrt(sq_arr)
print(sq_arr)
print(sqrt_arr) 
```

###단계 2: 비교 연산

```python
Comp = a > 2        # ①
print(Comp)         # ②
```

## 집계 함수: sum, mean, max, min, std


###단계 1: 기본 집계

```python
import numpy as np
data = np.array([[1, 2, 3], [4, 5, 6]]) # ①
# 전체 합, 평균
total_sum = np.sum(data)
mean_val = np.mean(data)
# 최대, 최소, 표준편차
max_val = np.max(data) # ②
min_val = np.min(data)
std_val = np.std(data)
print(total_sum)
print(mean_val)
print(max_val)
print(min_val)
print(std_val) 
```

###단계 2: 축별 집계

```python
col_sum = np.sum(data, axis=0)   # ①
row_mean = np.mean(data, axis=1) # ②
print(col_sum)
print(row_mean)
```

```python
import numpy as np
data = np.array([
[1, 2, 3],
[4, 5, 6]
]) # ①
max_idx_flat = np.argmax(data) # ②
min_idx_flat = np.argmin(data)
print(max_idx_flat)
print(min_idx_flat)
col_max_idx = np.argmax(data, axis=0) # ③
col_min_idx = np.argmin(data, axis=0)
print(col_max_idx)
print(col_min_idx)
row_max_idx = np.argmax(data, axis=1) # ④
row_min_idx = np.argmin(data, axis=1)
print(row_max_idx)
print(row_min_idx) 
```

## 인덱싱과 슬라이싱: 기본 슬라이싱, Boolean Indexing, Fancy Indexing

###단계 1: 기본 슬라이싱

```python
import numpy as np
data = np.array([[10, 20, 30],[40, 50, 60],[70, 80, 90]])        # ①
row_slice = data[1, :]              # ②
col_slice = data[:, 2]
print(row_slice)
print(col_slice)
print(data[0:2, 1:3])               # ③
```

###단계 2: Boolean Indexing

```python
bool_idx = data > 50  # ①
print(bool_idx)
print(data[bool_idx]) # ②
```

```python
scores = np.array([65, 82, 45, 91, 78, 55]) 
result = np.where(scores >= 60, "합격", "불합격")  # ①
print(result)
cleaned = np.where(scores < 70, 0, scores)       # ②
print(cleaned)
```

```python
import numpy as np
scores = np.array([[85, 92, 78],[60, 45, 88],[95, 100, 65],[30, 75, 55],[88, 60, 92]])# ①
print(scores)
low_avg_mask = np.mean(scores, axis=1) < 70 # ②
print(scores[low_avg_mask])
scores[low_avg_mask, 1] = np.where(low_avg_mask[low_avg_mask], 0, scores[low_avg_mask, 1])
print(scores)
grade = np.where(scores >= 90, "A+", scores.astype(str)) # ③
print(grade)
mask_kor_low = scores[:, 0] < 40
student_mean = np.mean(scores, axis=1, keepdims=True)
scores = np.where(mask_kor_low[:, np.newaxis], 
student_mean, scores) # ④
print(scores) 
```

```python
import numpy as np
scores = np.array([  # ①
   [85, 92, 78],
   [60, 45, 88],
   [95,100, 65],
   [30, 75, 55],
   [88, 60, 92]
])   # shape: (5, 3)
mean1 = np.mean(scores, axis=1)
print(mean1.shape)
mean2 = np.mean(scores, axis=1, keepdims=True)  # ②
print(mean2.shape)
mask = scores[:, 0] < 40
scores = np.where(mask[:, np.newaxis], mean2, scores)  # ③
print(mask[:, np.newaxis].shape)
print(mean2.shape)
print(scores.shape)
```

## 차원변환(reshape, transpose, flatten)

###단계 1: reshape

```python
arr = np.arange(12)                # ①
reshaped = arr.reshape(3, 4)       # ②
reshaped_auto = arr.reshape(3, -1) # ③
reshaped_flat = arr.reshape(-1, 6) # ④
print(reshaped)
print(reshaped_auto)
print(reshaped_flat)
```

###단계 2: transpose

```python
transposed = reshaped.T  # ①
print(transposed)
```

###단계 3: flatten

```python
flattened = reshaped.flatten()     # ①
flattened_t = transposed.flatten() # ②
print(flattened)
print(flattened_t)
```

연습 예시1: 주어진 데이터셋을 ndarray 로 변환하고 첫번째 열의 값을 2배로 만들고 전체 평균을 계산하시오.


```python
import numpy as np

data = [[10, 20, 35, 3], [1, 2, 3, 4]]
np_data = np.array(data)
np_data[:, 0] = np_data[:, 0] * 2
print(np_data)
mean_val = np.mean(np_data)
print(mean_val)
```

연습 예시2: 주어진 판매 데이터 리스트 [100, 200, 300]을 배열로 변환해 평균을 계산하시오.


```python
import numpy as np
sales_data = [100, 200, 300]
sales_data_arr = np.array(sales_data)
sales_average = np.mean(sales_data_arr)

print(sales_average)
```

연습 예시6: 주어진 데이터셋 값들을 2배로 한 후, 5 이상인 값들을 필터링하세요.


```python
data = np.array([1, 2, 3, 4])

doubled = data * 2
filtered = doubled[doubled > 5]

print(filtered)
```

연습 예시7: 주어진 2D 데이터셋에서 열 평균 계산 후, 평균 초과 행을 추출하세요.


```python
data = np.array([[1, 2], [3, 4], [5, 6]])
col_means = np.mean(data, axis=0)
exceed = data[(data[:, 0] > col_means[0]) & (data[:, 1] > col_means[1])]
print(exceed)
```

연습 예시8: 주어진 배열에서 짝수만 추출하세요.

연습 예시9: 주어진 배열에서 10보다 큰 값만 추출하세요.

연습 예시10: 주어진 2D 배열에서 첫 번째 행과 세 번째 행, 두 번째 열과 네 번째 열의 교차 요소를 추출하세요.

연습 예시11: 1~12까지 숫자를 담은 1D 배열을 3x4 배열로 변환하고, 전치(transpose) 후 1D로 평탄화(flatten) 하세요.

```python
arr = np.arange(1,13)
reshaped = arr.reshape(3,4)
transposed = reshaped.T
flattened = transposed.flatten()

print(reshaped)
print(transposed)
print(flattened)
```
