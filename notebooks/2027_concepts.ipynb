# 여섯 가지 핵심 개념

원고의 추가 개념도에 쓰인 작은 예시입니다. 데이터 파일 없이 실행할 수 있습니다. 실제 고객·모델 성능을 나타내는 자료가 아닙니다.

```python
import numpy as np
import pandas as pd
print('NumPy', np.__version__, '| pandas', pd.__version__)

from sklearn.preprocessing import OneHotEncoder
```

## 브로드캐스팅

![브로드캐스팅](https://raw.githubusercontent.com/Lilcob/AICE_Associate/main/assets/concepts/broadcasting.png)

```python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
vector = np.array([10, 20, 30])
print(matrix + vector)
print('결과 shape:', (matrix + vector).shape)
try:
    matrix + np.array([10, 20])
except ValueError:
    print('(2, 3)과 (2,)는 마지막 차원이 호환되지 않습니다.')
```

## loc와 iloc

![loc와 iloc](https://raw.githubusercontent.com/Lilcob/AICE_Associate/main/assets/concepts/loc_iloc.png)

```python
df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]}, index=['x', 'y', 'z'])
print(df.loc['y':'z'])
print(df.iloc[1:2])
```

## 키 기준 병합

![키 기준 병합](https://raw.githubusercontent.com/Lilcob/AICE_Associate/main/assets/concepts/merge_keys.png)

```python
left = pd.DataFrame({'ID': [101, 102, 103], '이름': ['가', '나', '다']})
right = pd.DataFrame({'ID': [102, 103, 104], '사용량': [20, 30, 40]})
print(left.merge(right, on='ID', how='inner', validate='one_to_one'))
print(left.merge(right, on='ID', how='left', validate='one_to_one'))
```

## 원핫 인코딩

![원핫 인코딩](https://raw.githubusercontent.com/Lilcob/AICE_Associate/main/assets/concepts/onehot_columns.png)

```python
train = pd.DataFrame({'색': ['red', 'blue', 'green']})
valid = pd.DataFrame({'색': ['yellow']})
encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
print(encoder.fit_transform(train))
print(encoder.get_feature_names_out())
print(encoder.transform(valid))
```

## 출력층과 손실함수

![출력층과 손실함수](https://raw.githubusercontent.com/Lilcob/AICE_Associate/main/assets/concepts/output_loss.png)

정수 클래스 라벨은 sparse_categorical_crossentropy, 원핫 라벨은 categorical_crossentropy로 같은 클래스 확률에 대한 손실을 계산할 수 있습니다. 아래는 확률이 [0.2, 0.3, 0.5]이고 정답이 클래스 2인 계산 예시입니다.

```python
probabilities = np.array([0.2, 0.3, 0.5])
integer_label = 2
onehot_label = np.array([0, 0, 1])
sparse_loss = -np.log(probabilities[integer_label])
categorical_loss = -np.sum(onehot_label * np.log(probabilities))
print(sparse_loss, categorical_loss)
assert np.isclose(sparse_loss, categorical_loss)
```

## 학습 곡선 읽기

![학습 곡선 읽기](https://raw.githubusercontent.com/Lilcob/AICE_Associate/main/assets/concepts/learning_curves.png)

위 그림은 경향을 설명하는 개념도입니다. 실제 학습 결과가 아닙니다. 훈련 손실만 줄어드는지, 검증 손실도 함께 줄어드는지 확인합니다. 조기 종료는 설정한 patience 동안의 개선 여부를 감시합니다.

```python
validation_loss = [0.8, 0.6, 0.5, 0.52, 0.54, 0.51]
best_epoch = int(np.argmin(validation_loss)) + 1
print('이 작은 예시의 검증 손실 최저 에포크:', best_epoch)
print('patience=3, min_delta=0이라면 최저점 이후 3회 미개선은 종료 조건이 됩니다.')
```
